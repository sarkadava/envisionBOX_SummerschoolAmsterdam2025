## Info
The current code allows you to train your own labeled videos, train a convolutional neural network that takes in body, hand, and face features from mediapipe holistic, and outputs behavioral classifications. Then you can apply the CNN to classify new videos with annotated data saved to csv's and the videos labeled for checking. Convolutional neural networks are great for classifying patterns in spatial and temporal data patterns. In this case we use CNN's to take a bunch of features of hand,body,face and to detect certain behaviors. To summarize, this code does the following:

* extract features with Mediapipe Holistic for training and inference
* train a CNN based on labels of the training data 'trainingvideos' folder
* loop over video snippets produce inferences on new videos in the 'videos_to_label' folder
* save the labeled videos and automatic annotations (in csvs) to the output folder

To code your own behaviors. Just change the gesture in the config code chunk below, and make sure that your training videos start with the name of the label. I advice that the videos are minimally as long as your window of inference (here 25 frames).

## Current code
The basic routine overviewed here is mostly from the NoddingPigeon code provided here: https://github.com/bhky/nodding-pigeon/. I have adapted it quite a bit, to make it more versatile for recognition of other types of gestures, and made the code completely linear, so that it is usable end-to-end. I have modified the following things:

* instead of taking simple features from the 2D head tracking module from mediapipe, we take the richer information of mediapipe holistic to get body, head, and hand information
* I have added head rotations derived from mediapipe as they are important for head nodding and turning
* I made the CNN model a bit more complex with more layers (i.e., a deeper network), to account for the more complex feature set, and added regularization and maxpooling (which should help with overfitting issues)
* further I have made the code such that you just run it from beginning to end, adapting a good bit to accomodate the linear working style that we use for envisionBOX

Please feel free to help out and improve the code. I will change the names in the citation of this module accordingly if you like. The current dataset is quite small, but I was still impressed with the accuracy for the three gestures, so im curious to know with more data how it performs on other types of data (see concluding remarks at the end).

# Setting up the config
The config file is very important and needs to be align with the training videos labels that you have. The gesture labels contain your behaviors and match your video names. When extracting the features for your training data, it will look for videos in 'the trainingvideos' folder with those names leading the filenames. 

```python
# set up the configuation
from dataclasses import dataclass
from typing import Tuple
from noddingpigeon import __weights_version__
import os


@dataclass
class Config:
    gesture_labels: Tuple[str, ...] = ("nodding", "turning", "clapping") #your gesture labelse here
    undefined_gesture_label: str = "undefined"
    stationary_label: str = "stationary"
    npz_filename: str = "./training/bodyheadlandmarks.npz" # this is where the training data is stored (mediapipe holistic body from the training videos)
    seq_length: int = 25 # this sets the window size for the classifier, so 25 frames input (so the model needs 25 frames to make a prediction)
    num_original_features: int = 29 # number of features that we originally take in (shown below, be careful this number should match you Feature set)
    weights_filename: str = f"./training/multi_gesture_weights_v0.1.1.weights.h5" # this contains the CNN model weights

Config = Config()

# set up the paths
curdir = os.getcwd()
trainingvideosfolder = os.path.abspath(os.path.join(curdir, "trainingvideos"))
print(f"Training videos folder: {trainingvideosfolder}")
```

# FEATURE EXTRACTION
So it is difficult to judge what features are crucial for your classification. Therefore I made the featureset now quite big, proposing features that I think are important for at least detecting hand gestures, head nods and head turning/shakes, and maybe shoulder and eye brow movement. The combined feature set is quite comprehensive. Of course, depending on your behavior you could add even more information about the hands, or the body. Most features comprise so called pose-embeddings, where we characterize some kind of movement of a joint relative to some other joint and normalize by something indicative of the body size. Below the following features are taken into account.

Overview of the features: 
1. rotation_x (pitch) (in degrees) -> e.g., important for nodding
2. rotation_y (yaw) (in degrees) -> e.g., important for turning
3. rotation_z (roll) (in degrees) -> e.g., important for tilting head
4. nose_x (position in pixel) -> e.g., important for general translational head movement
5. nose y (position in pixel) -> e.g., important for general translational head movement
6. nose z (position in pixel)  -> e.g., important for general translational head movement
7. distance nose to chin (normalizer) (in normalized units) -> used mostly to normalize the other distances
8. distance left eye brow to left eye center (in normalized units) -> e.g., important for eye brow movement
9. distance right eye brow to right eye center (in normalized units) -> e.g., important for eye brow movement
10. distance mouth corners (in normalized units) -> e.g., important for smiling
11. distance mouth apperture (in normalized units) -> e.g., important for opening mouth
12. distance left wrist to right wrist (in normalized units) -> e.g., important for hand gestures
13. distance left elbow to right elbow (in normalized units) -> e.g., important for hand gestures
14. distance left elbow to shoulder center (in normalized units) -> e.g., important for hand gestures
15. distance right elbow to shoulder center (in normalized units)   -> e.g., important for hand gestures
16. distance left wrist to shoulder center (in normalized units) -> e.g., important for hand gestures
17. distance right wrist to shoulder center (in normalized units) -> e.g., important for hand gestures
18. distance left shoulder to left ear (in normalized units)  -> e.g., important for shoulder shrugging
19. distance right shoulder to right ear (in normalized units) -> e.g., important for shoulder shrugging
20. distance left thumb to left index (in normalized units) -> e.g., important for hand gestures
21. distance right thumb to right index (in normalized units) -> e.g., important for hand gestures
22. distance left thumb to left pinky (in normalized units) -> e.g., important for hand gestures
23. distance right thumb to right pinky (in normalized units) -> e.g., important for hand gestures
24. distance x left wrist to x left elbow (in normalized units) -> e.g., important for hand gestures
25. distance x right wrist to x right elbow (in normalized units)   -> e.g., important for hand gestures
26. distance y left wrist to y left elbow (in normalized units) -> e.g., important for hand gestures
27. distance y right wrist to y right elbow (in normalized units) -> e.g., important for hand gestures
28. distance left index finger to nose (in normalized units) -> e.g., important for face touching
29. distance right index finger to nose (in normalized units) -> e.g., important for face touching

Now please note that the number of features need to be the same as in the config file. And your CNN model that we later define also needs to be adapted relative to the number of features. So if you want to make your life easier, for now dont change them, if you change them, keep this amount of features or adapt the script accordingly to accomodate your original feature numbers.

```python
"""
Video utilities.
"""
from enum import auto, Enum
from typing import List, Optional, Union

import cv2
import numpy as np
from mediapipe.python.solutions import drawing_utils as mp_drawing
from mediapipe.python.solutions import face_detection as mp_face
import mediapipe as mp #mediapipe
mp_holistic = mp.solutions.holistic

################## NAMES FOR MEDIAPIPE

markerxyzbody = [] #body landmarks
markerxyzhands = [] #hand landmarks
markerxyzface = [] #face landmarks

#landmarks 33x that are used by Mediapipe (Blazepose)
markersbody = ['NOSE', 'LEFT_EYE_INNER', 'LEFT_EYE', 'LEFT_EYE_OUTER', 'RIGHT_EYE_OUTER', 'RIGHT_EYE', 'RIGHT_EYE_OUTER',
          'LEFT_EAR', 'RIGHT_EAR', 'MOUTH_LEFT', 'MOUTH_RIGHT', 'LEFT_SHOULDER', 'RIGHT_SHOULDER', 'LEFT_ELBOW', 
          'RIGHT_ELBOW', 'LEFT_WRIST', 'RIGHT_WRIST', 'LEFT_PINKY', 'RIGHT_PINKY', 'LEFT_INDEX', 'RIGHT_INDEX',
          'LEFT_THUMB', 'RIGHT_THUMB', 'LEFT_HIP', 'RIGHT_HIP', 'LEFT_KNEE', 'RIGHT_KNEE', 'LEFT_ANKLE', 'RIGHT_ANKLE',
          'LEFT_HEEL', 'RIGHT_HEEL', 'LEFT_FOOT_INDEX', 'RIGHT_FOOT_INDEX']

markershands = ['LEFT_WRIST', 'LEFT_THUMB_CMC', 'LEFT_THUMB_MCP', 'LEFT_THUMB_IP', 'LEFT_THUMB_TIP', 'LEFT_INDEX_FINGER_MCP',
              'LEFT_INDEX_FINGER_PIP', 'LEFT_INDEX_FINGER_DIP', 'LEFT_INDEX_FINGER_TIP', 'LEFT_MIDDLE_FINGER_MCP', 
               'LEFT_MIDDLE_FINGER_PIP', 'LEFT_MIDDLE_FINGER_DIP', 'LEFT_MIDDLE_FINGER_TIP', 'LEFT_RING_FINGER_MCP', 
               'LEFT_RING_FINGER_PIP', 'LEFT_RING_FINGER_DIP', 'LEFT_RING_FINGER_TIP', 'LEFT_PINKY_FINGER_MCP', 
               'LEFT_PINKY_FINGER_PIP', 'LEFT_PINKY_FINGER_DIP', 'LEFT_PINKY_FINGER_TIP',
              'RIGHT_WRIST', 'RIGHT_THUMB_CMC', 'RIGHT_THUMB_MCP', 'RIGHT_THUMB_IP', 'RIGHT_THUMB_TIP', 'RIGHT_INDEX_FINGER_MCP',
              'RIGHT_INDEX_FINGER_PIP', 'RIGHT_INDEX_FINGER_DIP', 'RIGHT_INDEX_FINGER_TIP', 'RIGHT_MIDDLE_FINGER_MCP', 
               'RIGHT_MIDDLE_FINGER_PIP', 'RIGHT_MIDDLE_FINGER_DIP', 'RIGHT_MIDDLE_FINGER_TIP', 'RIGHT_RING_FINGER_MCP', 
               'RIGHT_RING_FINGER_PIP', 'RIGHT_RING_FINGER_DIP', 'RIGHT_RING_FINGER_TIP', 'RIGHT_PINKY_FINGER_MCP', 
               'RIGHT_PINKY_FINGER_PIP', 'RIGHT_PINKY_FINGER_DIP', 'RIGHT_PINKY_FINGER_TIP']

for mark in markersbody:
    for pos in ['X', 'Y', 'Z', 'visibility']: #for markers of the body you also have a visibility reliability score
        nm = pos + "_" + mark
        markerxyzbody.append(nm)
        
#check if there are numbers in a string
def num_there(s):
    return any(i.isdigit() for i in s)

#take some google classification object and convert it into a string
def makegoginto_str(gogobj):
    gogobj = str(gogobj).strip("[]")
    gogobj = gogobj.split("\n")
    return(gogobj[:-1]) #ignore last element as this has nothing

#make the stringifyd position traces into clean numerical values
def listpostions(newsamplemarks):
    newsamplemarks = makegoginto_str(newsamplemarks)
    tracking_p = []
    for value in newsamplemarks:
        if num_there(value):
            stripped = value.split(':', 1)[1]
            stripped = stripped.strip() #remove spaces in the string if present
            tracking_p.append(stripped) #add to this list  
    return(tracking_p)

class VideoSegment(Enum):
    BEGINNING = auto()
    LAST = auto()

## these features are given above
class Feature(Enum):
    rot_x = auto() #1 rotation x
    rot_y = auto() #2 rotation y
    rot_z = auto() #3 rotation z
    nose_x = auto() #4 nose
    nose_y = auto() #5 nose
    nose_z = auto() #6 nose
    norm_dist = auto() #7 distance from chin to top head
    left_brow_left_eye_norm_dist = auto() #8 left brow left eye distance
    right_brow_right_eye_norm_dist = auto() #9 right brow right eye distance
    mouth_corners_norm_dist = auto() #10 mouth corners distance
    mouth_apperture_norm_dist = auto() #11 mouth apperture distance
    left_right_wrist_norm_dist = auto() #12 left right wrist distance
    left_right_elbow_norm_dist = auto() #13 left right elbow distance
    left_elbow_midpoint_shoulder_norm_dist = auto() #14 left elbow midpoint shoulder distance
    right_elbow_midpoint_shoulder_norm_dist = auto() #15 right elbow midpoint shoulder distance
    left_wrist_midpoint_shoulder_norm_dist = auto() #16 left wrist midpoint shoulder distance
    right_wrist_midpoint_shoulder_norm_dist = auto() #17 right wrist midpoint shoulder distance
    left_shoulder_left_ear_norm_dist = auto() #18 left shoulder left ear distance
    right_shoulder_right_ear_norm_dist = auto() #19 right shoulder right ear distance
    left_thumb_left_index_norm_dist = auto() #20 left thumb left index distance
    right_thumb_right_index_norm_dist = auto() #21 right thumb right index distance
    left_thumb_left_pinky_norm_dist = auto() #22 left thumb left pinky distance
    right_thumb_right_pinky_norm_dist = auto() #23 right thumb right pinky distance
    x_left_wrist_x_left_elbow_norm_dist = auto() #24 x left wrist x left elbow distance
    x_right_wrist_x_right_elbow_norm_dist = auto() #25 x right wrist x right elbow distance
    y_left_wrist_y_left_elbow_norm_dist = auto() #26 y left wrist y left elbow distance
    y_right_wrist_y_right_elbow_norm_dist = auto() #27 y right wrist y right elbow distance
    left_index_finger_nose_norm_dist = auto() #28 left index finger nose distance
    right_index_finger_nose_norm_dist = auto() #29 right index finger nose distance
    
def video_to_landmarks(
        video_path: Optional[Union[int, str]],
        max_num_frames: Optional[int] = Config.seq_length,
        video_segment: VideoSegment = VideoSegment.BEGINNING,
        end_padding: bool = True,
        drop_consecutive_duplicates: bool = False
) -> List[List[float]]:
    assert Config.num_original_features == len(Feature)
    assert video_segment in VideoSegment
    video_path = video_path if video_path else 0  # For 0, webcam will be used.
    font_scale = 0.6
    line_thickness = 1

    valid_frame_count = 0
    prev_features: List[float] = []
    landmarks: List[List[float]] = []
    cap = cv2.VideoCapture(video_path)  # pylint: disable=no-member

    # lets extract the face landmarks, body, and hand
    with mp_holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic:
        while cap.isOpened():
            ret, bgr_frame = cap.read()
            if not ret:
                if video_path == 0:
                    continue  # Ignore empty frame of webcam.
                break  # End of given video.
            if max_num_frames and video_segment == VideoSegment.BEGINNING \
                    and valid_frame_count >= max_num_frames:
                break

            frame = cv2.cvtColor(bgr_frame, cv2.COLOR_BGR2RGB)  # pylint: disable=no-member
            # also do the same for holistic
            resultsh = holistic.process(frame)
            
            ######## extract the angles and the facemesh info
            h, w, c = frame.shape
            face_3d = []
            face_2d = []
            if  np.all(resultsh.face_landmarks) != None: #check if there is a pose found
                #R for the head we also estimate rotations
                for idx, lm in enumerate(resultsh.face_landmarks.landmark):
                    if idx == 33 or idx == 263 or idx == 1 or idx == 61 or idx == 291 or idx == 199:
                        if idx == 1:
                            nose_2d = (lm.x * w, lm.y * h)
                            nose_3d = (lm.x * w, lm.y * h, lm.z * 3000)
                        x, y = int(lm.x * w), int(lm.y * h)
                        face_2d.append([x, y])
                        face_3d.append([x, y, lm.z])
                #R Convert it to the NumPy array
                face_2d = np.array(face_2d, dtype=np.float64)
                #R Convert it to the NumPy array
                face_3d = np.array(face_3d, dtype=np.float64)  
                #R camera matrix
                focal_length = 1*w
                cam_matrix =  np.array([ [focal_length, 0, h / 2],
                                [0, focal_length, w / 2],
                                [0, 0, 1]])
                #R The distortion parameters
                dist_matrix = np.zeros((4, 1), dtype=np.float64)
                #R Solve the PnP problem
                success, rot_vec, trans_vec = cv2.solvePnP(face_3d, face_2d, cam_matrix, dist_matrix)
                #R Get rotational matrix
                rmat, jac = cv2.Rodrigues(rot_vec)              
                #R Get angles in normalized units
                angles, mtxR, mtxQ, Qx, Qy, Qz = cv2.RQDecomp3x3(rmat)
                # Get the rotation degree
                xrot = angles[0] * w # up and down
                yrot = angles[1] * h # left and right
                zrot = angles[2] * 3000 # tilt
                rotations = [xrot, yrot, zrot]

                # also get the nose
                nose_x = nose_3d[0]
                nose_y = nose_3d[1]
                nose_z = nose_3d[2]

                # get the normalized distance from chin to top head
                for idx, lm in enumerate(resultsh.face_landmarks.landmark):
                    if idx == 152 or idx == 10:
                        chin = (lm.x * w, lm.y * h)
                    if idx == 10:
                        top_head = (lm.x * w, lm.y * h)
                norm_dist = np.linalg.norm(np.array(chin) - np.array(top_head))

                # get the left and brow left eye distance
                for idx, lm in enumerate(resultsh.face_landmarks.landmark):
                    if idx == 133 or idx == 33:
                        left_inner_eye = (lm.x * w, lm.y * h)
                    if idx == 225:
                        left_brow = (lm.x * w, lm.y * h)
                left_brow_left_eye_norm_dist = np.linalg.norm(np.array(left_inner_eye) - np.array(left_brow))

                # get the right and brow right eye distance
                for idx, lm in enumerate(resultsh.face_landmarks.landmark):
                    if idx == 362 or idx == 263:
                        right_inner_eye = (lm.x * w, lm.y * h)
                    if idx == 225:
                        right_brow = (lm.x * w, lm.y * h)
                right_brow_right_eye_norm_dist = np.linalg.norm(np.array(right_inner_eye) - np.array(right_brow))

                # get the mouth corners distance
                for idx, lm in enumerate(resultsh.face_landmarks.landmark):
                    if idx == 87:
                        left_mouth = (lm.x * w, lm.y * h)
                    if idx == 308:
                        right_mouth = (lm.x * w, lm.y * h)
                mouth_corners_norm_dist = np.linalg.norm(np.array(left_mouth) - np.array(right_mouth))

                # get the mouth apperture distance
                for idx, lm in enumerate(resultsh.face_landmarks.landmark):
                    if idx == 13:
                        upper_lip = (lm.x * w, lm.y * h)
                    if idx == 14:
                        lower_lip = (lm.x * w, lm.y * h)
                mouth_apperture_norm_dist = np.linalg.norm(np.array(upper_lip) - np.array(lower_lip))

            # get the body points results.pose_landmarks based on markershands idx 
                # index, thumb, pinky, WRIST, ELBOW, SHOULDERS                              
                for idx, lm in enumerate(resultsh.pose_landmarks.landmark):
                    if idx == markersbody.index('LEFT_INDEX'):
                        left_index = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_INDEX'):
                        right_index = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('LEFT_THUMB'):
                        left_thumb = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_THUMB'):
                        right_thumb = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('LEFT_PINKY'):
                        left_pinky = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_PINKY'):
                        right_pinky = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('LEFT_WRIST'):
                        left_wrist = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_WRIST'):
                        right_wrist = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('LEFT_ELBOW'):
                        left_elbow = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_ELBOW'):
                        right_elbow = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('LEFT_SHOULDER'):
                        left_shoulder = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_SHOULDER'):
                        right_shoulder = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('LEFT_EAR'):
                        left_ear = (lm.x * w, lm.y * h)
                    if idx == markersbody.index('RIGHT_EAR'):
                        right_ear = (lm.x * w, lm.y * h)
                                    
                # get the distance between the left and right wrist normalized by norm_dist
                left_right_wrist_norm_dist = np.linalg.norm(np.array(left_wrist) - np.array(right_wrist)) / norm_dist
                # get the distance between the left and right elbow normalized by norm_dist
                left_right_elbow_norm_dist = np.linalg.norm(np.array(left_elbow) - np.array(right_elbow)) / norm_dist
                # get the distance between the left elbow and the midpoint of the shoulders normalized by norm_dist
                left_elbow_midpoint_shoulder_norm_dist = np.linalg.norm(np.array(left_elbow) - np.array(left_shoulder)) / norm_dist
                # get the distance between the right elbow and the midpoint of the shoulders normalized by norm_dist
                right_elbow_midpoint_shoulder_norm_dist = np.linalg.norm(np.array(right_elbow) - np.array(right_shoulder)) / norm_dist
                # get the distance between the left wrist and the midpoint of the shoulders normalized by norm_dist
                left_wrist_midpoint_shoulder_norm_dist = np.linalg.norm(np.array(left_wrist) - np.array(left_shoulder)) / norm_dist
                # get the distance between the right wrist and the midpoint of the shoulders normalized by norm_dist
                right_wrist_midpoint_shoulder_norm_dist = np.linalg.norm(np.array(right_wrist) - np.array(right_shoulder)) / norm_dist
                # get the distance between the left shoulder and the left ear normalized by norm_dist
                left_shoulder_left_ear_norm_dist = np.linalg.norm(np.array(left_shoulder) - np.array(left_ear)) / norm_dist
                # get the distance between the right shoulder and the right ear normalized by norm_dist
                right_shoulder_right_ear_norm_dist = np.linalg.norm(np.array(right_shoulder) - np.array(right_ear)) / norm_dist
                # get the distance between the left thumb and the left index normalized by norm_dist
                left_thumb_left_index_norm_dist = np.linalg.norm(np.array(left_thumb) - np.array(left_index)) / norm_dist
                # get the distance between the right thumb and the right index normalized by norm_dist
                right_thumb_right_index_norm_dist = np.linalg.norm(np.array(right_thumb) - np.array(right_index)) / norm_dist
                # get the distance between the left thumb and the left pinky normalized by norm_dist
                left_thumb_left_pinky_norm_dist = np.linalg.norm(np.array(left_thumb) - np.array(left_pinky)) / norm_dist
                # get the distance between the right thumb and the right pinky normalized by norm_dist
                right_thumb_right_pinky_norm_dist = np.linalg.norm(np.array(right_thumb) - np.array(right_pinky)) / norm_dist
                # get the distance between the x left wrist and the x left elbow normalized by norm_dist
                x_left_wrist_x_left_elbow_norm_dist = (left_wrist[0] - left_elbow[0]) / norm_dist
                # get the distance between the x right wrist and the x right elbow normalized by norm_dist
                x_right_wrist_x_right_elbow_norm_dist = (right_wrist[0] - right_elbow[0]) / norm_dist
                # get the distance between the y left wrist and the y left elbow normalized by norm_dist
                y_left_wrist_y_left_elbow_norm_dist = (left_wrist[1] - left_elbow[1]) / norm_dist
                # get the distance between the y right wrist and the y right elbow normalized by norm_dist
                y_right_wrist_y_right_elbow_norm_dist = (right_wrist[1] - right_elbow[1]) / norm_dist
                # get the distance between the left index finger and the nose normalized by norm_dist
                left_index_finger_nose_norm_dist = np.linalg.norm(np.array(left_index) - np.array(nose_2d)) / norm_dist
                # get the distance between the right index finger and the nose normalized by norm_dist
                right_index_finger_nose_norm_dist = np.linalg.norm(np.array(right_index) - np.array(nose_2d)) / norm_dist

                # collect the features into as we have just created them above
                    # this feature set has 29 features
                features = [xrot, yrot, zrot, nose_x, nose_y, nose_z, norm_dist, 
                            left_brow_left_eye_norm_dist, right_brow_right_eye_norm_dist, 
                            mouth_corners_norm_dist, mouth_apperture_norm_dist, 
                            left_right_wrist_norm_dist, left_right_elbow_norm_dist, 
                            left_elbow_midpoint_shoulder_norm_dist, right_elbow_midpoint_shoulder_norm_dist, 
                            left_wrist_midpoint_shoulder_norm_dist, right_wrist_midpoint_shoulder_norm_dist, 
                            left_shoulder_left_ear_norm_dist, right_shoulder_right_ear_norm_dist, 
                            left_thumb_left_index_norm_dist, right_thumb_right_index_norm_dist, 
                            left_thumb_left_pinky_norm_dist, right_thumb_right_pinky_norm_dist, 
                            x_left_wrist_x_left_elbow_norm_dist, x_right_wrist_x_right_elbow_norm_dist, 
                            y_left_wrist_y_left_elbow_norm_dist, y_right_wrist_y_right_elbow_norm_dist, 
                            left_index_finger_nose_norm_dist, right_index_finger_nose_norm_dist]
                
                assert len(features) == Config.num_original_features

                if drop_consecutive_duplicates and np.array_equal(
                        np.round(features, decimals=2),
                        np.round(prev_features, decimals=2)
                ):
                    continue

                landmarks.append(features)
                prev_features = features

                valid_frame_count += 1
            else:
                detection = None

            if video_path != 0:
                continue

            if detection:
                mp_drawing.draw_detection(frame, detection)

            flipped_frame = cv2.flip(frame, 1)  # pylint: disable=no-member

            text = f"{valid_frame_count} / {max_num_frames}" \
                if max_num_frames else str(valid_frame_count)
            cv2.putText(  # pylint: disable=no-member
                flipped_frame, text, (10, 20),
                cv2.FONT_HERSHEY_SIMPLEX,  # pylint: disable=no-member
                font_scale, (0, 0, 255),
                line_thickness, cv2.LINE_AA  # pylint: disable=no-member
            )

            cv2.imshow(  # pylint: disable=no-member
                "Webcam",
                cv2.cvtColor(flipped_frame, cv2.COLOR_RGB2BGR)  # pylint: disable=no-member
            )

            if cv2.waitKey(1) == ord("q"):  # pylint: disable=no-member
                break

        cap.release()
        cv2.destroyAllWindows()  # pylint: disable=no-member

        if not landmarks:
            return []

        if max_num_frames and video_segment == VideoSegment.LAST:
            landmarks = landmarks[-max_num_frames:]

        if max_num_frames and end_padding and len(landmarks) < max_num_frames:
            last = landmarks[-1]
            landmarks = landmarks + [last] * (max_num_frames - len(landmarks))

        return landmarks


#if __name__ == "__main__":
    #video_to_landmarks(None, None)

print('Setting up functions for feature extraction done, please proceeed with setting up the CNN model')
```

# Setting up feature preprocessing and setting up your CNN architecture
The following code sets up the CNN architecture that you are going to use. Preprocessing creates the input data, a tensor of the features, the time derivatives of the features (for capturing change), and their standard deviations. Then these are fed into convolutional layers that take some window size and extract higher order features using filters. The first layer has 32 filters, second 64 filters, and the third layer 128 filters. Pooling reduces the output to a layer with a lower dimensionality, which is then flattened into a 1D vector. Dropout and regularizers are to prevent overfitting. We have two outputs, has_motion = for overal motion detection, and gesture_probs = for the probablities of the gestures.

```python
"""
Model utilities.
"""
import os
from typing import Dict, List, Optional

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, regularizers # pylint: disable=import-error
from tensorflow.keras.models import Model  # pylint: disable=import-error,no-name-in-module

#from noddingpigeon.config import Config
from noddingpigeon._download import download_weights_to, get_default_weights_path

import importlib.util
import sys

def make_model(weights_path: Optional[str] = get_default_weights_path()) -> Model:
    seq_input = layers.Input(
        shape=(Config.seq_length, Config.num_original_features),
        dtype=tf.float32, name="input"
    )
    x = seq_input
    x = Preprocessing()(x)
    # Shape: (batch_size, seq_length, num_features)
    x = layers.Conv1D(32, 3, strides=1, padding="same", activation="relu")(x)
    x = layers.MaxPooling1D(pool_size=2, strides=1, padding='same')(x)
    x = layers.Conv1D(64, 3, strides=1, padding="same", activation="relu")(x)
    x = layers.MaxPooling1D(pool_size=2, strides=1, padding='same')(x)
    x = layers.Conv1D(128, 3, strides=1, padding="same", activation="relu", kernel_regularizer=regularizers.l2(0.01))(x)
    x = layers.MaxPooling1D(pool_size=2, strides=1, padding='same')(x)
    x = layers.Flatten()(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.5)(x)
    x_0 = layers.Dropout(0.01)(x)

    has_motion = layers.Dense(1, activation="sigmoid", name="has_motion")(x_0)
    gesture_probs = layers.Dense(len(Config.gesture_labels),
                                 activation="softmax", name="gesture_probs")(x_0)
    output = layers.Concatenate()([has_motion, gesture_probs])

    model = Model(seq_input, output)

    if weights_path is not None:
        if not os.path.isfile(weights_path):
            download_weights_to(weights_path)
        model.load_weights(weights_path)

    return model

def load_landmarks(npz_path: str) -> Dict[str, List[List[float]]]:
    loaded = np.load(npz_path)
    return {label: loaded[label].tolist() for label in loaded.files}


class Preprocessing(layers.Layer):  # type: ignore

    def __init__(self) -> None:
        super(Preprocessing, self).__init__(name="preprocessing")

    @staticmethod
    def call(x: tf.Tensor,
            training: Optional[bool] = None,
            mask: Optional[tf.Tensor] = None
    ) -> tf.Tensor:
        features = x

        # we center all the features
        features = features - tf.reduce_mean(features, axis=-2, keepdims=True)
        
        # we take derivative of the features
        t_deriv = features[:, 1:] - features[:, :-1]

        # we pad with a zero vector
        t_deriv = tf.pad(t_deriv, [[0, 0], [1, 0], [0, 0]])

        # then we take the standard deviation of the features
        coord_std = tf.math.reduce_std(features, axis=-2)
        t_deriv_std = tf.math.reduce_std(t_deriv, axis=-2)

        # Concatenate new features and exclude unwanted ones.
        features = tf.concat(  # pylint: disable=unexpected-keyword-arg,no-value-for-parameter
            [features, t_deriv], axis=-1
        )

        return features
```

# Training
### Note: this can be skipped if you already trained a model, but do make sure you have run the other chunks so that you can do the inference
We now have now extracted the features of the training data, which will be used for - indeed - training the CNN classifier on the categories. Note that the training can take some time, but on a CPU it does not take me longer than 30 mins. You can speed it up by making sure you use tensorflow-gpu. 

```python
# import losses
from tensorflow.keras import losses

tf.random.set_seed(0)


def setup_accelerators_and_get_strategy() -> tf.distribute.Strategy:
    os.environ["CUDA_VISIBLE_DEVICES"] = "0,1"
    gpu_devices = tf.config.list_physical_devices("GPU")
    if gpu_devices:
        # Strategy for GPU or multi-GPU machines.
        strategy = tf.distribute.MirroredStrategy()
        print(f"Number of accelerators: {strategy.num_replicas_in_sync}")
    else:
        strategy = tf.distribute.OneDeviceStrategy("/cpu:0")
        print("Using single CPU.")
    return strategy


def make_y(label_idx: int) -> List[int]:
    has_motion = 1 if label_idx > 0 else 0
    y = [has_motion] + [0] * len(Config.gesture_labels)
    if has_motion == 1:
        y[label_idx] = 1
    # Note:
    # Format of y is [has_motion, <one-hot-gesture-class-vector>], e.g.,
    # [1, 0, 1, ..., 0] for the 2nd gesture,
    # [0, ..., 0] for stationary case.
    return y


def make_ds_train(
        landmark_dict: Dict[str, Sequence[Sequence[float]]],
        seq_length: int,
        num_features: int,
        seed: int
) -> tf.data.Dataset:
    # Note: stationary label must come first in this design, see make_y.
    labels = (Config.stationary_label,) + Config.gesture_labels
    rng = np.random.default_rng(seed=seed)

    def gen() -> Tuple[List[List[float]], int]:
        while True:
            label_idx = int(rng.integers(len(labels), size=1))
            landmarks = landmark_dict[labels[label_idx]]
            seq_idx = int(rng.integers(len(landmarks) - seq_length, size=1))
            features = landmarks[seq_idx: seq_idx + seq_length]
            yield features, make_y(label_idx)

    return tf.data.Dataset.from_generator(
        gen,
        output_signature=(
            tf.TensorSpec(shape=(seq_length, num_features), dtype=tf.float32),
            tf.TensorSpec(shape=(len(labels),), dtype=tf.int32)
        )
    )


def loss(y_true: tf.Tensor, y_pred: tf.Tensor) -> tf.Tensor:
    has_motion_true = y_true[:, :1]
    has_motion_pred = y_pred[:, :1]
    has_motion_loss = losses.BinaryCrossentropy(
        reduction=tf.keras.losses.Reduction.NONE
    )(has_motion_true, has_motion_pred)

    # The gesture loss is designed in the way that, if has_motion is 0,
    # the box values do not matter.
    mask = y_true[:, 0] == 1
    weight = tf.where(mask, 1.0, 0.0)
    gesture_true = y_true[:, 1:]
    gesture_pred = y_pred[:, 1:]
    gesture_loss = losses.CategoricalCrossentropy(
        label_smoothing=0.05,
        reduction=tf.keras.losses.Reduction.NONE
    )(gesture_true, gesture_pred, sample_weight=weight)

    return (has_motion_loss + gesture_loss) * 0.5


class CustomAccuracy(tf.keras.metrics.Metric):

    def __init__(
            self,
            motion_threshold: float = 0.5,
            gesture_threshold: float = 0.9,
            name: str = "custom_accuracy"
    ) -> None:
        super(CustomAccuracy, self).__init__(name=name)
        self.motion_threshold = motion_threshold
        self.gesture_threshold = gesture_threshold
        self.acc = tf.keras.metrics.CategoricalAccuracy()

    def update_state(
            self,
            y_true: tf.Tensor,
            y_pred: tf.Tensor,
            sample_weight: Optional[tf.Tensor] = None
    ) -> None:
        # IMPORTANT - the sample_weight parameter is needed to solve:
        # TypeError: tf__update_state() got an unexpected keyword argument 'sample_weight'
        y_pred = tf.where(y_pred[:, :1] >= self.motion_threshold, y_pred, 0.0)
        y_pred = tf.where(y_pred >= self.gesture_threshold, 1.0, 0.0)
        self.acc.update_state(y_true[:, 1:], y_pred[:, 1:])

    def result(self) -> tf.Tensor:
        return self.acc.result()

    def reset_state(self) -> None:
        self.acc.reset_state()


def compile_model(model: Model) -> None:
    model.compile(
        loss=loss,
        optimizer=tf.keras.optimizers.Adam(amsgrad=True),
        metrics=[CustomAccuracy()],
    )


def get_steps_per_epoch(
        landmark_dict: Dict[str, Sequence[Sequence[float]]]
) -> int:
    # Kind of arbitrary here.
    mean_data_size = int(np.mean([len(v) for v in landmark_dict.values()]))
    steps_per_epoch = int(mean_data_size * 0.7)
    return steps_per_epoch


def train_and_save_weights(
        landmark_dict: Dict[str, List[List[float]]],
        model: Model,
        weights_path: str,
        seed: int = 42
) -> None:
    ds_train = make_ds_train(
        landmark_dict, Config.seq_length, Config.num_original_features, seed
    )
    ds_train = ds_train.batch(16).prefetch(tf.data.AUTOTUNE)

    callbacks = [
        tf.keras.callbacks.ModelCheckpoint(
            filepath=weights_path, monitor="loss", mode="min",
            save_best_only=True, save_weights_only=True, verbose=1
        ),
        tf.keras.callbacks.EarlyStopping(
            monitor="loss", min_delta=1e-04, patience=10,
            restore_best_weights=True, verbose=1
        ),
    ]
    model.fit(
        ds_train,
        epochs=500,
        steps_per_epoch=get_steps_per_epoch(landmark_dict),
        callbacks=callbacks,
        verbose=1,
    )


def main() -> None:
    strategy = setup_accelerators_and_get_strategy()
    with strategy.scope():
        model = make_model(weights_path=None)
        compile_model(model)
    landmark_dict = load_landmarks(Config.npz_filename)
    try:
        train_and_save_weights(landmark_dict, model, Config.weights_filename)
    except KeyboardInterrupt:
        print("Training interrupted.")


if __name__ == "__main__":
    main()
```

# Set up the inferencing code
This code follows the pipeline in pigeonnodding. It sets up the decision boundaries, saves the results in a dictionary, and produces landmarks for the video to be used for inference on CNN model that we set up in the config.

```python
"""
Inference utilities.
"""
from dataclasses import asdict, dataclass
from typing import Any, Dict, Optional, Sequence

import numpy as np
from tensorflow.keras.models import Model  # pylint: disable=import-error,no-name-in-module

@dataclass
class Result:
    gesture: str
    probabilities: Dict[str, Any]

def postprocess(
        prediction: Sequence[float],
        motion_threshold: float,
        gesture_threshold: float
) -> Dict[str, Any]:
    if not prediction:
        return asdict(Result(Config.undefined_gesture_label, {}))

    gesture_probs = prediction[1:]
    if prediction[0] < motion_threshold:
        label = Config.stationary_label
    else:
        if np.max(gesture_probs) < gesture_threshold:
            label = Config.undefined_gesture_label
        else:
            label = Config.gesture_labels[int(np.argmax(gesture_probs))]
    result = Result(
        gesture=label,
        probabilities={
            "has_motion": prediction[0],
            "gestures": {
                Config.gesture_labels[i]: gesture_probs[i]
                for i in range(len(Config.gesture_labels))
            }
        }
    )
    return asdict(result)


def predict_video(
        video_path: Optional[str] = None,  # None will start a webcam.
        model: Optional[Model] = None,
        max_num_frames: int = Config.seq_length,  # For the pre-trained model.
        video_segment: VideoSegment = VideoSegment.BEGINNING,
        end_padding: bool = True,
        drop_consecutive_duplicates: bool = True,
        postprocessing: bool = True,
        motion_threshold: float = 0.5,
        gesture_threshold: float = 0.9
) -> Any:
    if model is None:
        model = make_model()
    landmarks = video_to_landmarks(
        video_path, max_num_frames, video_segment, end_padding,
        drop_consecutive_duplicates
    )

    if landmarks:
        prediction: Sequence[float] = model.predict(
            np.expand_dims(landmarks, axis=0)
        )[0].tolist()
    else:
        prediction = []

    if not postprocessing:
        return prediction
    return postprocess(prediction, motion_threshold, gesture_threshold)


#if __name__ == "__main__":
#     print(predict_video(model=make_model(f"../training/{Config.weights_filename}")))
```

# Applying your model to the to be labeled videos (final step)
The below code show applies the predictions at a window of 25 frames (but you could set this at the start of your pipeline in the config) with 250ms window, so we get labels every 250 milliseconds. The labels are then stored in the output folder, with annotated videos. Now the code just prints the label that has the highest likelihood. But the csv data show all the likelihoods of the gestures. In some cases you could of course have multiple gestures.

Note that you can change the likelihood that is needed before you treat something as a gesture, there first need to be motion (now set at 0.5 prob) for a gesture to be considered there. A gesture will be labeled as such when the probability is higher than .95. You can change these settings of course to make the model more sensitive. 

```python
# Inference on videos and save the results - without subclip
import tempfile
import cv2
import pandas as pd
import numpy as np
import os
import glob

def get_video_property(cap, prop_name):
    """Get video property with backwards compatibility for older OpenCV versions"""
    # Try modern constants first
    if hasattr(cv2, 'CAP_PROP_FPS'):
        prop_mapping = {
            'fps': cv2.CAP_PROP_FPS,
            'frame_count': cv2.CAP_PROP_FRAME_COUNT,
            'frame_width': cv2.CAP_PROP_FRAME_WIDTH,
            'frame_height': cv2.CAP_PROP_FRAME_HEIGHT,
            'pos_frames': cv2.CAP_PROP_POS_FRAMES
        }
        return cap.get(prop_mapping[prop_name])
    # Try older constants
    elif hasattr(cv2, 'cv') and hasattr(cv2.cv, 'CV_CAP_PROP_FPS'):
        prop_mapping = {
            'fps': cv2.cv.CV_CAP_PROP_FPS,
            'frame_count': cv2.cv.CV_CAP_PROP_FRAME_COUNT,
            'frame_width': cv2.cv.CV_CAP_PROP_FRAME_WIDTH,
            'frame_height': cv2.cv.CV_CAP_PROP_FRAME_HEIGHT,
            'pos_frames': cv2.cv.CV_CAP_PROP_POS_FRAMES
        }
        return cap.get(prop_mapping[prop_name])
    # Fall back to numeric constants
    else:
        prop_mapping = {
            'fps': 5,
            'frame_count': 7,
            'frame_width': 3,
            'frame_height': 4,
            'pos_frames': 1
        }
        return cap.get(prop_mapping[prop_name])

def set_video_property(cap, prop_name, value):
    """Set video property with backwards compatibility"""
    if hasattr(cv2, 'CAP_PROP_FPS'):
        prop_mapping = {
            'pos_frames': cv2.CAP_PROP_POS_FRAMES
        }
        return cap.set(prop_mapping[prop_name], value)
    elif hasattr(cv2, 'cv') and hasattr(cv2.cv, 'CV_CAP_PROP_FPS'):
        prop_mapping = {
            'pos_frames': cv2.cv.CV_CAP_PROP_POS_FRAMES
        }
        return cap.set(prop_mapping[prop_name], value)
    else:
        prop_mapping = {
            'pos_frames': 1
        }
        return cap.set(prop_mapping[prop_name], value)
def extract_frames_sequence(video_path, start_time, duration, target_fps=None):
    """
    Extract a sequence of frames from video between start_time and start_time+duration
    
    Args:
        video_path: Path to video file
        start_time: Start time in seconds
        duration: Duration in seconds
        target_fps: Target FPS for extraction (None = original FPS)
    
    Returns:
        List of frames as numpy arrays
    """
    cap = cv2.VideoCapture(video_path)
    original_fps = get_video_property(cap, 'fps')
    
    if target_fps is None:
        target_fps = original_fps
    
    # Calculate frame numbers
    start_frame = int(start_time * original_fps)
    end_frame = int((start_time + duration) * original_fps)
    
    frames = []
    frame_count = 0
    
    # Set to start frame
    set_video_property(cap, 'pos_frames', start_frame)
    
    while frame_count < (end_frame - start_frame):
        ret, frame = cap.read()
        if not ret:
            break
            
        frames.append(frame)
        frame_count += 1
    
    cap.release()
    return frames

def save_frames_as_video(frames, output_path, fps):
    """
    Save a list of frames as a video file
    """
    if not frames:
        return False
        
    height, width, channels = frames[0].shape
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))
    
    for frame in frames:
        out.write(frame)
    
    out.release()
    return True

# Main processing loop
augmented_video_fol = os.path.join('.', 'output', 'labeled_videos')
automated_annotation_fol = os.path.join('.', 'output', 'automated_annotations')
all_raw_videos_fol = os.path.join('.', 'videos_to_label')

# Create output directories if they don't exist
os.makedirs(augmented_video_fol, exist_ok=True)
os.makedirs(automated_annotation_fol, exist_ok=True)

all_raw_videos = glob.glob(os.path.join(os.path.abspath(all_raw_videos_fol), '*.mp4'))

# identify the model
print(Config.weights_filename)
MODEL = make_model(Config.weights_filename)

# Process each video
for vidnam in all_raw_videos:
    # get the clean name without path and extension
    vidnamcl = os.path.basename(vidnam)
    vidnamcl = vidnamcl.split('.')[0]
    
    # Get video properties using OpenCV - backwards compatible
    cap = cv2.VideoCapture(vidnam)
    frame_rate = get_video_property(cap, 'fps')
    frame_count = int(get_video_property(cap, 'frame_count'))
    durationvideo = frame_count / frame_rate
    cap.release()
    
    print(f'duration of video is: {durationvideo} seconds')
    print(f'frame rate of video is: {frame_rate} frames per second')
    
    # Set up sliding window parameters
    windowframestime = Config.seq_length / frame_rate
    subclipstart = 0
    subclipend = windowframestime
    increment = 0.25
    results = []
    
    # Process video sections for gesture annotation
    print(f'PHASE1: detecting head gestures for: {vidnamcl}')
    
    while subclipend < durationvideo:
        # Extract frames for this time window
        frames = extract_frames_sequence(vidnam, subclipstart, windowframestime)
        
        if frames:
            # Create temporary video file from frames
            temp_dir = tempfile.mkdtemp()
            temp_video_path = os.path.join(temp_dir, 'temp.mp4')
            
            # Save frames as temporary video
            if save_frames_as_video(frames, temp_video_path, frame_rate):
                # Predict gesture in this window
                result = predict_video(temp_video_path, 
                                     motion_threshold=0.7, 
                                     gesture_threshold=0.95, 
                                     model=MODEL)
                results.append(result)
            else:
                # Handle case where video creation failed
                results.append({
                    'gesture': 'unknown',
                    'probabilities': {
                        'has_motion': 0.0,
                        'gestures': {'nodding': 0.0, 'turning': 0.0}
                    }
                })
            
            # Clean up temporary file
            if os.path.exists(temp_video_path):
                os.remove(temp_video_path)
            os.rmdir(temp_dir)
        
        print(f"{subclipstart:.2f} - {subclipend:.2f}")
        
        # Move to next window
        subclipstart += increment
        subclipend += increment

    print(f'done with detecting head gestures for: {vidnamcl}')

    # Create DataFrame with results
    rows = []
    for i, entry in enumerate(results):
        row = {
            'start': i * increment,
            'end': (i * increment) + windowframestime,
            'timemiddle': (i * increment) + (windowframestime/2),
            'gesture': entry['gesture'],
            'has_motion': entry['probabilities']['has_motion'],
            'nodding_prob': entry['probabilities']['gestures']['nodding'],
            'turning_prob': entry['probabilities']['gestures']['turning']
        }
        rows.append(row)
    
    df = pd.DataFrame(rows)
    
    # Save results to CSV
    csv_output_path = os.path.join(automated_annotation_fol, f'{vidnamcl}_auto_annotated.csv')
    df.to_csv(csv_output_path, index=False)
    
    print('the following results were obtained: ')
    print(df.head())
    print('PHASE2: now we are going to add the labels to the video')

    # Add labels to original video using OpenCV
    cap = cv2.VideoCapture(vidnam)
    frame_width = int(get_video_property(cap, 'frame_width'))
    frame_height = int(get_video_property(cap, 'frame_height'))
    
    outputvideo = os.path.join(augmented_video_fol, f'{vidnamcl}_auto_annotated.mp4')
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # More compatible than MJPG
    out = cv2.VideoWriter(outputvideo, fourcc, frame_rate, (frame_width, frame_height))
    
    # Process video frame by frame and add labels
    current_time = 0
    frame_number = 0
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        
        # Find corresponding gesture label for current time
        labels = df[(current_time >= df['start']) & (current_time <= df['end'])]
        
        if not labels.empty:
            # Get the label with the smallest time difference from current time
            time_diffs = abs((labels['start'] + labels['end']) / 2 - current_time)
            closest_idx = time_diffs.idxmin()
            gesture_label = labels.loc[closest_idx, 'gesture']
            
            # Add text overlay
            cv2.putText(frame, gesture_label, (40, 80), 
                       cv2.FONT_HERSHEY_SIMPLEX, 2, (255, 255, 255), 3, cv2.LINE_AA)
            cv2.putText(frame, gesture_label, (40, 80), 
                       cv2.FONT_HERSHEY_SIMPLEX, 2, (0, 0, 0), 2, cv2.LINE_AA)
        
        out.write(frame)
        frame_number += 1
        current_time = frame_number / frame_rate
    
    cap.release()
    out.release()
    cv2.destroyAllWindows()
    
    print(f'Completed processing: {vidnamcl}')
```
